# Set up database

```python
from dotenv import load_dotenv
import os

load_dotenv()

user = os.getenv("DB_USER")
password = os.getenv("DB_PASS")
host = os.getenv("DB_HOST")
dbname = os.getenv("DB_NAME")

from sqlalchemy import create_engine
engine = create_engine(f"mysql+mysqlconnector://{user}:{password}@{host}/{dbname}", connect_args={'init_command': 'SET time_zone="+07:00"'})
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
n_trials = 10      # Number of trials
p_success = 0.5    # Probability of success
n_samples = 1000   # Number of observations

# Generate binomial-distributed data
data = np.random.binomial(n=n_trials, p=p_success, size=n_samples)

# Create a DataFrame
df = pd.DataFrame(data, columns=['successes'])

# Count frequency of each outcome
counts = df['successes'].value_counts().sort_index()

# Display frequency table
print(counts)

# Optional: Plot the distribution
counts.plot(kind='bar', title='Binomial Distribution (n=10, p=0.5)')
plt.xlabel('Number of successes')
plt.ylabel('Frequency')
plt.show()
```
