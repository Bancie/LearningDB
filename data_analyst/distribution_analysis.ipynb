# Binomial probability distribution

```python
import pandas as pd
from scipy.stats import binom
import matplotlib.pyplot as plt
```

$n$ trials

```python
n = 10
```

$p$ success

```python
p = 0.5
```

table

```python
k_values = list(range(n + 1))

probabilities = [binom.pmf(k, n, p) for k in k_values]

df = pd.DataFrame({
    'Successes (k)': k_values,
    'P(X = k)': probabilities
})

print(df)
```

```python
df.plot(x='Successes (k)', y='P(X = k)', kind='bar', legend=False, title='Binomial Probability Distribution')
plt.ylabel('Probability')
plt.xlabel('Number of Successes')
plt.grid(True)
plt.show()
```
