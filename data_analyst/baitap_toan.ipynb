# Set up database

```python
from dotenv import load_dotenv
import os

load_dotenv()

user = os.getenv("DB_USER")
password = os.getenv("DB_PASS")
host = os.getenv("DB_HOST")
dbname = os.getenv("DB_NAME")

from sqlalchemy import create_engine
engine = create_engine(f"mysql+mysqlconnector://{user}:{password}@{host}/{dbname}", connect_args={'init_command': 'SET time_zone="+07:00"'})
```

# PHONE SCREEN

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
query = "SELECT * FROM `baitap_toan`"
ps = pd.read_sql(query, engine)
```

```python
ps.head()
```

```python
print(ps[['TOTAL_COUNT','minutes']].describe())
```

```python
ps.boxplot(column=['TOTAL_COUNT','minutes'])

plt.title('Boxplot of Screen time')
plt.ylabel('Minute')
plt.show()
```

```python
ps['TOTAL_COUNT'].plot.hist()
plt.xlabel("TOTAL_COUNT")
plt.title("Distribution of TOTAL_COUNT")
plt.show()
```

```python
ps['minutes'].plot.hist()
plt.xlabel("minutes")
plt.title("Distribution of minutes")
plt.show()
```
